##### Create a class Book with attributes title, author, and price.
##### Add a method discount(percent) that reduces the price by the given percentage.

```python
class Book:
    def __init__(self, title: str, author: str, price:float):
        self.title=title
        self.author=author
        self.price=price
    
    def discount(self,percent:float):
        self.price -= self.price * (percent/100)

book = Book("book1","foo", 100)
book.discount(10)
print(f"book orignal price 100 after discounted {book.price}")
```

##### Define a class Rectangle with methods to calculate area and perimeter.
##### Add a method to check if it’s a square.

```python
class Rectangle:
    
    def __init__(self,width:float, hight:float):
        self.width=width
        self.hight=hight
    
    def area(self) -> float:
        return self.width * self.hight 
    
    def perimeter(self):
        return (self.hight+self.width) * 2   
    
    def is_square(self) -> bool:
        return self.hight == self.width
    

rect = Rectangle(10,20)
print(f"Rectangle area {rect.area()}")
print(f"Rectangle perimeter {rect.perimeter()}")
print(f"Rectangle is suqared {rect.is_square()}")
```

##### Write a class Temperature that supports conversion between Celsius and Fahrenheit using class methods.

```python
class Temperature:
    
    @classmethod
    def convert_cles_Fahr(cls, cels:float) -> float:
        return (cels * 9/5) + 32
    
    @classmethod
    def convert_Fahr_cels(cls, fahr:float) -> float:
        return (fahr - 32) * 5/9
    
print(f"Temp converted form celsius to farh 35 -> {Temperature.convert_cles_Fahr(35)}")
print(f"Temp converted form farh to celcius 95.0 -> {Temperature.convert_Fahr_cels(95.0)}")

"""
Design a class ShoppingCart that uses a class variable to track total carts created.
"""
class ShoppingCart:
    totalCarts=0
    
    def __init__(self, cart_number:int=0):
        self.cart_number=cart_number
        ShoppingCart.totalCarts +=1
    
    @classmethod    
    def total_carts(cls):
        print(f"total carts created {cls.totalCarts}")
        

shopping_cart_101 =ShoppingCart(101)
shopping_cart_102 =ShoppingCart(102)
shopping_cart_103 =ShoppingCart(103)
ShoppingCart.total_carts()
```

##### 1) Create an abstract base class Shape with abstract method area(). Implement subclasses Circle and Rectangle

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area():
        pass

class Circle(Shape):
    def __init__(self, radius:float):
        self.radius=radius
        
    def area(self):
        print(f"area of circle is {3.14 * self.radius **2 }")

class Rectangle(Shape):
    def __init__(self, hight:float, width:float):
        self.hight=hight
        self.width=width
        
    def area(self):
        print(f"area of Rectangle is {(self.hight * self.width) }")


circle = Circle(10)
circle.area()

rectangle = Rectangle(10,20)
rectangle.area()
```

##### 2 ) Implement a class Vector that overloads +, -, and == operators. Test with multiple vector objects.

```python
class MyVector:

    def __init__(self, p1:float,p2:float,p3:float):
        self.p1=p1
        self.p2=p2
        self.p3=p3

    def __add__(self,v2:MyVector):
        print("calling add method")
        return MyVector(self.p1+v2.p1,self.p2+v2.p2,self.p3+v2.p3)

    def __sub__(self,v2:MyVector):
        print("calling substract method")
        return MyVector(self.p1-v2.p1,self.p2-v2.p2,self.p3-v2.p3)

    def __eq__(self,other:MyVector):
        print("Calling __eq__() method")
        return self.p1==other.p1 and self.p2==other.p2 and self.p3==other.p3

    def __repr__(self):
        return f"MyVector[p1={self.p1},p2={self.p2},p3={self.p3}]"

    
v1 = MyVector(1,2,3)
v2 = MyVector(4,5,6)
print(v1 + v2)
print(v1 - v2)
print(v1 == v2)
```

##### 3) Write a class Logger that acts as a context manager to log messages to a file using __enter__ and __exit__.

```python
class MyLogger:

    def __init__(self, filename):
        self.filename=filename
        self.file=None

    def __enter__(self):
        print("Open the file in append mode")
        self.file = open(self.filename, "a")
        return self

    def log_message(self, message:str):
        self.file.write(message+"\n")

    def __exit__(self,exc_type, exc_value, traceback):
        if self.file:
            print("file is closed")
            self.file.close()
        return False


with MyLogger("log.txt") as myLogger:
    myLogger.log_message("message1")
    myLogger.log_message("message2")
    myLogger.log_message("message3")

```

##### 4) Build a class Plugin with a class-level registry that automatically registers subclasses when defined.

```python
class MyPlugin:
    registry = {}
    def __init_subclass__(cls, **kwargs):
        print("Automatically called when a subclass is created.")
        super().__init_subclass__(**kwargs)
        MyPlugin.registry[cls.__name__]=cls

    def run(self):
        """Abstract method to be implemented by subclasses."""
        raise NotImplementedError("Subclasses must implement run()")

class DataCleaner(MyPlugin):
    def run(self):
        print("Cleaning data...")

class DataExporter(MyPlugin):
    def run(self):
        print("Exporting data...")

print("Registered plugins:", MyPlugin.registry.items())
for name,plugin_cls in MyPlugin.registry.items():
    print(f"Running {name}:")
    plugin_cls().run()
```

##### 1) Create a metaclass AutoDoc that automatically adds a docstring to every class created with it.

```python
class AutoDoc(type):

    def __new__(mcs,name,base,namespace):
        if "__doc__" not in namespace or not namespace["__doc__"]:
            namespace["__doc__"] = f"Auto-generated docstring for class {name}"
        return super().__new__(mcs,name,base,namespace)

# classes that extend autodoc
class MyClass(metaclass=AutoDoc):
    pass

class OtherClass(metaclass=AutoDoc):
    """This is doc string of other class"""
    pass

print(MyClass.__doc__)
print(OtherClass.__doc__)
```

##### 2) Design a class Workflow that uses composition — it should contain multiple Task objects and execute them sequentially.

```python
class Task:
    def __init__(self,name,action):
        self.name=name
        self.action=action

    def execute(self):
        print(f"Executing task: {self.name}")
        self.action()

class Workflow:

    def __init__(self, name):
        self.name=name
        self.task=[]

    def add_task(self, task):
        print("Adding task to workflow -> ", task.name)
        self.task.append(task)

    def run(self):
        print(f"starting workflow {self.name}")
        for task in self.task:
            task.execute()
        print(f"Workflow '{self.name}' completed.")

def clean_data():
    print("Cleaning data...")
def transform_data():
    print("Transforming data...")
def export_data():
    print("Exporting data...")

# Create tasks
t1 = Task("Data Cleaning", clean_data)
t2 = Task("Data Transformation", transform_data)
t3 = Task("Data Export", export_data)

# Build workflow
wf = Workflow("ETL Pipeline")
wf.add_task(t1)
wf.add_task(t2)
wf.add_task(t3)

# Run workflow
wf.run()
```

##### 3) Implement a class ConfigLoader that reads JSON configuration files and caches them using a static method.

```python
import json

class ConfigLoader:
    _cache= {}

    @staticmethod
    def load_config(filename):

        if filename in ConfigLoader._cache:
            print(f"found in cache return from cache {filename}")
            return ConfigLoader._cache[filename]
            
        with open(filename, "r") as file:
            print(f"load data from {filename}")
            data = json.load(file)

        ConfigLoader._cache[filename]=data

        return data

print(ConfigLoader.load_config('log.txt'))
print(ConfigLoader.load_config('log.txt'))
```
