##### Create a class Book with attributes title, author, and price.
##### Add a method discount(percent) that reduces the price by the given percentage.

```python
class Book:
    def __init__(self, title: str, author: str, price:float):
        self.title=title
        self.author=author
        self.price=price
    
    def discount(self,percent:float):
        self.price -= self.price * (percent/100)

book = Book("book1","foo", 100)
book.discount(10)
print(f"book orignal price 100 after discounted {book.price}")
```

##### Define a class Rectangle with methods to calculate area and perimeter.
##### Add a method to check if it’s a square.

```python
class Rectangle:
    
    def __init__(self,width:float, hight:float):
        self.width=width
        self.hight=hight
    
    def area(self) -> float:
        return self.width * self.hight 
    
    def perimeter(self):
        return (self.hight+self.width) * 2   
    
    def is_square(self) -> bool:
        return self.hight == self.width
    

rect = Rectangle(10,20)
print(f"Rectangle area {rect.area()}")
print(f"Rectangle perimeter {rect.perimeter()}")
print(f"Rectangle is suqared {rect.is_square()}")
```

##### Build a class Student that stores marks for three subjects. Add a method average() to compute the average score.

```python
class Student:
    
    def __init__(self, sub1:float=0.0, sub2:float=0.0, sub3:float=0.0,):
        self.sub1=sub1
        self.sub2=sub2
        self.sub3=sub3
        
    def avarge(self) -> float:
        return (self.sub1+self.sub2+self.sub3)/3.0
    
student = Student(10,20,30)
print(f"Student avg score {student.avarge()}")
```

##### Implement inheritance:
Base class Employee with attributes name and salary.
Subclass Manager that adds department and overrides a method display_info().

```python
class Employee:
    
    def __init__(self, name:str,salary:float):
        self.name=name
        self.salary=salary
    
    def display_info(self):
        print(f"I am employee {self.name} and my salary is {self.salary}")

class Manager(Employee):
    
    def display_info(self):
        print(f"I am manager {self.name} and my salary is {self.salary}")
        
employee = Employee("foo",1234)
manager = Manager("bar",4321)

employee.display_info()        
manager.display_info() 
```

##### Write a class Temperature that supports conversion between Celsius and Fahrenheit using class methods.

```python
class Temperature:
    
    @classmethod
    def convert_cles_Fahr(cls, cels:float) -> float:
        return (cels * 9/5) + 32
    
    @classmethod
    def convert_Fahr_cels(cls, fahr:float) -> float:
        return (fahr - 32) * 5/9
    
print(f"Temp converted form celsius to farh 35 -> {Temperature.convert_cles_Fahr(35)}")
print(f"Temp converted form farh to celcius 95.0 -> {Temperature.convert_Fahr_cels(95.0)}")

"""
Design a class ShoppingCart that uses a class variable to track total carts created.
"""
class ShoppingCart:
    totalCarts=0
    
    def __init__(self, cart_number:int=0):
        self.cart_number=cart_number
        ShoppingCart.totalCarts +=1
    
    @classmethod    
    def total_carts(cls):
        print(f"total carts created {cls.totalCarts}")
        

shopping_cart_101 =ShoppingCart(101)
shopping_cart_102 =ShoppingCart(102)
shopping_cart_103 =ShoppingCart(103)
ShoppingCart.total_carts()
```
